```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    r"C:\Users\Hiral\OneDrive\Desktop\AML_DeepLearning_Project\dataset\PS_20174392719_1491204439457_log.csv"
)
```

```python
df = df.drop(
    columns=["nameOrig", "nameDest"]
)
```

```python
print(df.columns.tolist())
```

```python
print(
    "Duplicates:",
    df.duplicated().sum()
)
```

```python
df = df.drop_duplicates()
```

```python
print(df.isnull().sum())
```

```python
numeric_columns = [
    "step",
    "amount",
    "oldbalanceOrg",
    "newbalanceOrig",
    "oldbalanceDest",
    "newbalanceDest"
]

for column in numeric_columns:
    print(
        column,
        "negative values:",
        (df[column] < 0).sum()
    )
```

```python
print(
    df["isFlaggedFraud"].value_counts()
)
```

```python
df = df.drop(
    columns=["isFlaggedFraud"]
)
```

```python
print(df.columns.tolist())
```

```python
df["balanceChangeOrg"] = (
    df["oldbalanceOrg"]
    - df["newbalanceOrig"]
)
```

```python
df["balanceChangeDest"] = (
    df["newbalanceDest"]
    - df["oldbalanceDest"]
)
```

```python
df["amountToBalanceRatio"] = (
    df["amount"] /
    (df["oldbalanceOrg"] + 1)
)
```

```python
print(df.columns.tolist())
```

```python
df.to_csv(
    r"C:\Users\Hiral\OneDrive\Desktop\AML_DeepLearning_Project\dataset\final_features.csv",
    index=False
)
```
